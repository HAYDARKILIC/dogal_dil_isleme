## 1. Önerilen proje kulvarları

1. kontrol noktasında, onaya tabi olmak üzere kendi kulvarınızı önerebilirsiniz. Aşağıda önceden incelenmiş üç kulvarın hepsi, zaman bütçesi içinde güçlü bir teslime olanak tanır.

### Kulvar A — Alan ön eğitimi
Küçük, açık bir dil modelinin (ör. 100–300M parametreli bir model) ön eğitimini odaklı bir alan derlemi üzerinde sürdürün — PubMed özetleri, hukuki içtihatlar, finansal raporlar, bilimsel kod. Ayrılmış bir alan kümesinde perpleksiteyi ve tek bir alan kıyaslamasında alt görev kazanımlarını ölçün.

**Zorluk:** orta. **Hesaplama:** 4 saatlik bir sürdürülen ön eğitim için tek bir tüketici sınıfı GPU yeterlidir.

### Kulvar B — Düşük kaynaklı dil
Yeterli kaynağı olmayan bir dil (Kürtçe, Azerbaycan Türkçesi, Özbekçe, Keçuva, ...) için sıfırdan bir tokenizer ve küçük bir dil modeli eğitin. Tek bir alt görevde değerlendirin — duygu sınıflandırma, adlandırılmış varlık tanıma veya basit makine çevirisi.

**Zorluk:** daha yüksek (darboğaz veri toplamadır). **Hesaplama:** küçük.

### Kulvar C — Yetenek değerlendirme paketi
Seçtiğiniz tek bir yetenek için titiz bir değerlendirme paketi kurun — matematiksel akıl yürütme, kod üretimi, olgusallık veya talimat izleme. Üç açık modeli bu paketle kıyaslayın. Güven aralıklarını raporlayın. Mümkün olduğunda kirlenmeyi tespit edin.

**Zorluk:** daha düşük teknik taban, daha yüksek yöntem çıtası. **Hesaplama:** API veya yalnızca çıkarım.

Kendi veri kümenizi veya alt görevinizi getirmeniz teşvik edilir — en ilginç projelerin çoğu standart kıyaslamaların dışından gelir.

## 5. Kaçınılması gereken anti-desenler

Önceki dönemlerdeki yaygın hatalara dayanan kısa bir saha rehberi.

- **Taban çizgisi yok.** "Modelim %73 doğruluğa ulaşıyor." Karşılaştırma sayısı olmadan bu yorumlanamaz. Her zaman en az bir basit taban çizgisi ekleyin (çoğunluk sınıfı, BM25, eğitilmemiş rastgele başlatma, ayarlanmamış referans model).
- **Güven aralığı yok.** Küçük değerlendirme kümelerinde tek sayılık farklar genellikle gürültüdür. Bootstrap uygulayın (Hafta 11) veya tohum başına varyansı raporlayın.
- **Sızıntı.** Eğitim bölümüyle örtüşen ya da hiperparametre seçiminde kullanılmış bir test bölümü her şeyi geçersiz kılar. Eğitim/geliştirme/test bölümlerini herhangi bir modellemeden *önce* tanımlayın.
- **Seçilmiş niteliksel örnekler.** Üç iyi çıktı hiçbir şey kanıtlamaz. Üretilen metin gösteriyorsanız düzgün rastgele örnekleyin.
- **Ablasyon yok.** "X'i denedim ve işe yaradı." Sonucu hangi bileşen sağladı? Doğal "bileşeni çıkar" deneyini yapın.
- **Hata analizi yok.** Hata örnekleri içermeyen bir rapor eksiktir. Bir makalenin en ilginç sayfası genellikle hata analizidir.

## 6. Araçlar ve kaynaklar

Başvurabileceğiniz pratik kaynakların eksiksiz olmayan bir listesi.

**Veri.** Hugging Face Datasets, Common Crawl (filtrelenmiş), dile özgü derlemler (ör. Türkçe için TS Corpus), alana özgü dökümler (PubMed, S2ORC, The Stack).

**Kayıt tutma.** Weights & Biases, MLflow veya düz bir CSV — bir hafta sonra neyi çalıştırdığınızı yeniden kurmanızı sağlayan herhangi bir şey. Tekrarlanabilirlik bir notlandırma ölçütüdür.
